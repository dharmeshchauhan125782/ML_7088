```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Convert horsepower to numeric, replacing any non-numeric value (like '?') with NaN
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
# Fill any missing value in horsepower with the median value
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].median())

# Select the available numeric columns from the dataset
x = df[['engine-size', 'compression-ratio', 'horsepower', 'city-mpg', 'highway-mpg']]

# Rescale data
scaler = MinMaxScaler()
rescaled_data = scaler.fit_transform(x)

# Display result
print(rescaled_data)
```

```python
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder object
encoder = LabelEncoder()

# Encode categorical column
df['make'] = encoder.fit_transform(df['make'])

# Display encoded data
print(df[['make']].head())
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Ensure 'df' is loaded in the active runtime
try:
    _ = df.head()
except NameError:
    try:
        df = pd.read_csv('/content/Automobile_data.csv')
    except FileNotFoundError:
        url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data'
        column_names = ['symboling', 'normalized-losses', 'make', 'fuel-type', 'aspiration', 'num-of-doors', 'body-style', 'drive-wheels', 'engine-location', 'wheel-base', 'length', 'width', 'height', 'curb-weight', 'engine-type', 'num-of-cylinders', 'engine-size', 'fuel-system', 'bore', 'stroke', 'compression-ratio', 'horsepower', 'peak-rpm', 'city-mpg', 'highway-mpg', 'price']
        df = pd.read_csv(url, names=column_names)

# Create LabelEncoder object
encoder = LabelEncoder()

# Encode categorical column
df['price'] = encoder.fit_transform(df['price'].astype(str))

# Display encoded data
print(df[['price']].head())
```
